   - IAM & Admin → Service Accounts → Create → Download JSON
   - Share your Google Sheets with the service account email

### Troubleshooting:
- **"No profiles found"**: ICP too restrictive, broaden criteria
- **"Permission denied"**: Share Google Sheets with service account email
- **"Insufficient credits"**: Add credits to Apify/OpenRouter accounts
- **"Invalid credentials"**: Check that you replaced all placeholders in Step 3
